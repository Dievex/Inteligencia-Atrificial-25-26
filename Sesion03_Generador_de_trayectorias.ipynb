# Generador de Trayectorias
### Esto es un generador de trayectorias p2p

v1: Generación de linea recta entre 2 puntos

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the grid size
grid_size = 100

# Get user input for start and end points as tuples
start_input = input(f"Enter the starting coordinates (x,y) (0-{grid_size-1}): ")
end_input = input(f"Enter the ending coordinates (x,y) (0-{grid_size-1}): ")

# Parse the input tuples
start_x, start_y = map(int, start_input.strip("()").split(","))
end_x, end_y = map(int, end_input.strip("()").split(","))


# Create an empty grid
grid = np.zeros((grid_size, grid_size))

# Implement Bresenham's line algorithm
x0, y0 = start_x, start_y
x1, y1 = end_x, end_y

dx = abs(x1 - x0)
dy = abs(y1 - y0)
sx = 1 if x0 < x1 else -1
sy = 1 if y0 < y1 else -1
err = dx - dy

points = []
while True:
    points.append((x0, y0))
    if x0 == x1 and y0 == y1:
        break
    e2 = 2 * err
    if e2 > -dy:
        err -= dy
        x0 += sx
    if e2 < dx:
        err += dx
        y0 += sy

# Mark the points on the grid
for x, y in points:
    if 0 <= x < grid_size and 0 <= y < grid_size:
        grid[y, x] = 1 # Note: grid indexing is typically (row, column) or (y, x)

# Display the grid
plt.imshow(grid, cmap='binary', origin='lower')
plt.title("Generated Line on Grid")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.show()
```
