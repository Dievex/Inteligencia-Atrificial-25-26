# Generador de Trayectorias
### Esto es un generador de trayectorias p2p

v1: Generación de linea recta entre 2 puntos

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# Define the grid size
grid_size = 100

# Get user input for start and end points as tuples
start_input = input(f"Enter the starting coordinates (x,y) (0-{grid_size-1}): ")
end_input = input(f"Enter the ending coordinates (x,y) (0-{grid_size-1}): ")

# Parse the input tuples
start_x, start_y = map(int, start_input.strip("()").split(","))
end_x, end_y = map(int, end_input.strip("()").split(","))

# Create an empty grid
grid = np.zeros((grid_size, grid_size))

# Implement Bresenham's line algorithm to get the points
x0, y0 = start_x, start_y
x1, y1 = end_x, end_y

dx = abs(x1 - x0)
dy = abs(y1 - y0)
sx = 1 if x0 < x1 else -1
sy = 1 if y0 < y1 else -1
err = dx - dy

points = []
while True:
    points.append((x0, y0))
    if x0 == x1 and y0 == y1:
        break
    e2 = 2 * err
    if e2 > -dy:
        err -= dy
        x0 += sx
    if e2 < dx:
        err += dx
        y0 += sy

# Create the figure and axis for the animation
fig, ax = plt.subplots()
ax.set_xlim(0, grid_size)
ax.set_ylim(0, grid_size)
ax.set_aspect('equal', adjustable='box')
ax.set_title("Generated Line and Robot Movement")
ax.set_xlabel("X-axis")
ax.set_ylabel("Y-axis")
ax.grid(True)

# Display the grid background
ax.imshow(grid, cmap='binary', origin='lower', extent=[0, grid_size, 0, grid_size])

# Plot the trajectory
path_x, path_y = zip(*points)
ax.plot(path_x, path_y, 'r-', linewidth=2) # Plot the path in red

# Create the robot marker
robot, = ax.plot([], [], 'bo', markersize=10) # Blue circle marker for the robot

# Animation update function
def update(frame):
    # Update the robot's position
    robot.set_data([points[frame][0]], [points[frame][1]])
    return robot,

# Create the animation
ani = FuncAnimation(fig, update, frames=len(points), blit=True, interval=100) # Adjust interval for speed

# Display the animation
plt.close(fig) # Close the figure to avoid displaying a static plot
display(HTML(ani.to_jshtml()))
```
